```python
#Return the given number rounded to two decimal places.
def round_to_two_places(num):
    return round(num,2)
num=3.456
print(round_to_two_places(num))
```

```python
#The help for round says that ndigits (the second argument) may be negative. What do you think will happen when it is? Try some examples in the following cell.
def round_number(num):
    return round(num,-2)
num=1234.7896
print(round_number(num))
```

```python
"""Return the number of leftover candies that must be smashed after distributing
    the given number of candies evenly between 3 friends."""
def to_smash(total_candies):
    print("Smashed: ",total_candies%3)
bob_candies=int(input("Enter total number of candies of bob"))
alice_candies=int(input("Enter total number of candies of alice"))
tom_candies=int(input("Enter total number of candies of tom"))
total_candies=bob_candies+alice_candies+tom_candies
print(to_smash(total_candies))
```
